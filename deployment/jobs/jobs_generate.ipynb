```python
import json
from dbruntime.databricks_repl_context import get_context
import requests

job_id = dbutils.widgets.get("job_id")
```

```python


host = spark.conf.get("spark.databricks.workspaceUrl")

token = dbutils.notebook.entry_point.getDbutils() \
    .notebook().getContext().apiToken().get()


headers = {
    "Authorization": f"Bearer {token}"
}

response = requests.get(
    f"https://{host}/api/2.2/jobs/get",
    headers=headers,
    params={"job_id": job_id}
)

job = response.json()
job_def = job["settings"]

# -----------------------
# get job name safely
# -----------------------
job_name = job_def.get("name", f"job_{job_id}")

# sanitize file name
job_name = job_name.replace(" ", "_").replace("/", "_")

# -----------------------
# write to workspace path
# -----------------------
file_path = f"/Workspace/Shared/jobs/{job_name}.json"

with open(file_path, "w") as f:
    json.dump(job_def, f, indent=2)

print(f"Saved job definition to: {file_path}")
```
